```python
import pandas as pd
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.describe(include = 'all')
```

```python
df.isnull().sum()
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
#create age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(30)
```

```python
# create row namely purchase_frequency_days

frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90}

df['purchase_frequency_days'] =  df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['discount_applied','promo_code_used']].head()
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
df = df.drop('promo_code_used', axis=1)
```

```python
from sqlalchemy import create_engine

server = r"localhost\SQLEXPRESS"
database = "customer_behavior"

connection_string = (
    f"mssql+pyodbc://@{server}/{database}"
    "?trusted_connection=yes&driver=ODBC+Driver+17+for+SQL+Server"
)

engine = create_engine(connection_string)

print("Connection created")
```

```python
with engine.connect() as conn:
    print("SQL Server connected successfully!")
```

```python
df.head()
```

```python
df.head(10)
```

```python
df.to_sql(
    "customer_behavior_cleaned",
    con=engine,
    if_exists="replace",
    index=False
)

print("Data loaded into SQL Server!")
```

```python
result = pd.read_sql(
    "SELECT TOP 10 * FROM customer_behavior_cleaned",
    engine
)

result
```
